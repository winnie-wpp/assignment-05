```python
# You make have noticed using the vanilla CSV Reader has some limitations
# So we're going to use a powerful Python library called Pandas
import pandas as pd  # in the Python community, pd is the agreed shorthand for pandas
# this will help us make graphs
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# read the csv file
path = 'data/pokemon_data.csv'
# in python 1 or more rows is called a Dataframe
# pandas naming conventions (ie best practices) have all DataFrame variables end in _df
pokemon_df = pd.read_csv(path)    # Fun Fact: the plural form of Pokemon is Pokemon
```

```python
# we can look at just the headers
pokemon_df.columns
```

```python
# base_happiness is a STRING, not a variable so it needs quotes!
# bracket notation
pokemon_df['base_happiness']
```

```python
# if a column has no spaces, you can use dot notation
pokemon_df.capture_rate
```

```python
display(pokemon_df)
```

```python
# let's look at the first five rows
pokemon_df.head()
```

```python
# we can use indexing with DataFrames
pokemon_df[12:19]   
```

```python
# A single column in pandas is referred to as a Series
pokemon_df.type1
```

```python
# we can also use indexing on a Series
pokemon_df.type1[0]
```

```python
# we can modify data in a DataFrame
pokemon_df.type1 = pokemon_df.type1.str.capitalize()
# now all type1 values will start with a upper case letter
# this is extremely helpful if you are trying to make all values the same type and/or 
# formatting
display(pokemon_df)
```

```python
# I can use the loc property to locate a row by a certain value
# for example, I could find the row for the pokemon 'Pikachu'
pokemon_df.loc[pokemon_df.name == 'Pikachu']
```

```python
# i can also search for rows by index
pokemon_df.iloc[24]
```

```python
# I can pull certain values from a row
pokemon_df.iloc[24].japanese_name
```

```python
# I could also use the loc property to just see the japanese name of the pokemon 'Pikachu'
# so the first parameter is my search query, the second parameter is the column value I want 
# to return
pokemon_df.loc[pokemon_df.name == 'Pikachu', 'japanese_name']
```

```python
# item function only works if you have 1 item in series
pokemon_df.loc[pokemon_df.name == 'Pikachu', 'japanese_name'].item()
```

```python
# the loc function returns all rows that match the condition
# this allows me to get all grass type pokemon
# remember I capitalized the type1 values, so I need to say 'Grass', not 'grass'
pokemon_df.loc[pokemon_df.type1 == 'Grass', 'name']
```

```python
# Analyze data with Pandas
# For example, I can see the highest and lowest values for HP
print(pokemon_df.hp.max())
print(pokemon_df.hp.min())
```

```python
# let's find out who these pokemon are!
# we're going to use the loc property to locate any pokemon where HP equals 255 or 1!
strong_pokemon = pokemon_df.loc[pokemon_df.hp == 255, 'name'].item()
weak_pokemon = pokemon_df.loc[pokemon_df.hp == 1, 'name'].item()
print(strong_pokemon)
print(weak_pokemon)
```

```python
# let's make a DataFrame of the strongest pokemon
strongest_pokemon_df = pokemon_df[pokemon_df.hp>125]
strongest_pokemon_df
```

```python
# I can look at their names
strongest_pokemon_df.name
```

```python
# let's see how many unique types there are among these pokemon
print(strongest_pokemon.type1.unique())
```

```python
# If bins is an integer, it defines the number of equal-width bins in the range.
plt.hist(pokemon_df.hp, bins=25)
plt.xlabel('HP')
plt.ylabel('Number of Pokemon')
plt.title('Distribution of HP amongst Pokemon')
plt.axvline(pokemon_df.hp.mean(), color='yellow', linestyle='dashed', linewidth=2)    # draw where the average is in yellow
plt.axvline(pokemon_df.hp.median(), color='red', linestyle='dashed', linewidth=2)  # draw where the median is in red
```

```python
# let's group our pokemon by type and see the differenes in HP
# so we are going to split our pokemon DataFrame by 'type1' using the groupby() function 
# then we'll calculate the median HP for each type
type_groupings = pokemon_df.groupby('type1')
type_groupings.hp.median()
```

```python
# we can also see a deeper analysis of these numbers by using describe()
# 68% Rule: In a bell-shaped normal distribution, approximately 68.3% of all data points fall 
# within one standard deviation above or below the mean.
# A score that sits exactly one standard deviation above the mean is at the 84th percentile, 
# while a score one standard deviation below the mean is at the 16th percentile.
# https://encrypted-tbn2.gstatic.com/images?q=tbn:ANd9GcQ5zQmpvtohc_KZnrOjrVPcd9q_CVs6RWB8MY-Lw7VCc_Rgqxee
type_groupings.hp.describe()
```

```python
# let's graph water vs fire

# make a dataframe of just water pokemon
water_df = type_groupings.get_group('Water')
# make a dataframe of just fire pokemon
fire_df = type_groupings.get_group('Fire')
plt.hist(water_df.hp, bins=20, label='Water')
plt.hist(fire_df.hp, bins=20, label='Fire')
plt.legend()
plt.title("Distribution of HP by type")
plt.xlabel("HP")
plt.ylabel("Number of Pokemon")
```

```python
# find the numerical correlation between two colums
# meaning: how does speed correlate to a pokemon's hp?
correlation = pokemon_df.speed.corr(pokemon_df.hp)

print(correlation) # answer: not very (1 is the strongest positive correlation, -1 strongest neg coor, 0 is no correlation)
```

```python
pokemon_values = pokemon_df[['speed', 'hp', 'sp_attack', 'sp_defense']]
corr_pokemon = pokemon_values.corr()
corr_pokemon
```

```python
plt.matshow(pokemon_values.corr())
plt.xticks(range(pokemon_values.select_dtypes(['number']).shape[1]), pokemon_values.select_dtypes(['number']).columns, fontsize=14, rotation=45)
plt.yticks(range(pokemon_values.select_dtypes(['number']).shape[1]), pokemon_values.select_dtypes(['number']).columns, fontsize=14)
cb = plt.colorbar()
cb.ax.tick_params(labelsize=14)
plt.title('Correlation Matrix', fontsize=16);
plt.show()

# we can see a stronger correlation bewteen special attack and special defense and a weaker
# correlation with hp and speed
```
